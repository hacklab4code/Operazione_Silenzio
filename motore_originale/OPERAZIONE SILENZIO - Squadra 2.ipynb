```python
import subprocess, sys, os
CHI = 'Persona 1'   # chi sta lavorando: finisce nel diario
def esegui(*argomenti):
    """Esegue  python operazione_silenzio.py <argomenti>  e stampa l'output riga per riga."""
    env = dict(os.environ, PYTHONUTF8='1', PYTHONIOENCODING='utf-8')
    p = subprocess.Popen([sys.executable, 'operazione_silenzio.py', *argomenti], stdout=subprocess.PIPE,
                         stderr=subprocess.STDOUT, text=True, encoding='utf-8', errors='replace', env=env)
    for riga in p.stdout: print(riga, end='')
    return p.wait()
assert os.path.exists('operazione_silenzio.py'), 'Aprite il notebook dalla cartella del progetto'
```
